```python
import pandas as pd
import matplotlib.pyplot as plt
```

# Basic


```python
# Q1.Data sheet /set load.
file = pd.read_csv("C:\\Users\\pavitr\\OneDrive\\Documents\\project\\Python Project\\sale_store_data.csv")
file
```

```python
# Q2. Head and info check.
h = file.head() # first 5 rows
h
```

```python
i = file.info() # strucuture
i
```

```python
# Q3. Missing values identify.
file.isnull().sum() # find null values 
```

# Cleaning 

```python
# Q4. Duplicate remove.
file = file.drop_duplicates() # remove duplicate.
```

```python
# Q5. Data type correct.
file.dtypes # check data type.
```

```python
# Q6. Date column format check.
df = file
df['Order Date'] = pd.to_datetime(df['Order Date']) # fix date.
df.head()
```

# Analysis

```python
# Q7. Total sales and profit.
S = df['Sales'].sum()  # Total sales.
P = df['Profit'].sum() # Total Profit.
print("This is Total Sales:-" ,S,"This is Total Profit:-",P )
```

```python
# Q8.  Category wise sales.
df.groupby('Category')['Sales'].sum() # Category-wise sales
```

```python
# Q9. Sub category wise profit.
df.groupby('Sub-Category')['Profit'].sum() # Sub-category profit
```

```python
# Q10. Top 10 customers. 
df.groupby('Customer Name')['Sales'].sum().sort_values(ascending=False).head(10)
```

# Advanced

```python
# Q11. Monthy Sales Trend.
# Monthly sales
df['Month'] = df['Order Date'].dt.month
df.groupby('Month')['Sales'].sum()
```

```python

# Q13. Highest selling product.
# Highest selling product
df.groupby('Product Name')['Sales'].sum().sort_values(ascending=False).head(1)
```

```python
# Q14. Lowest profit profuct.
# Lowest profit product
df.groupby('Product Name')['Profit'].sum().sort_values().head(1)
```

# Visualization

```python
# Q15.Bar chart(Category sales).
# Bar chart
df.groupby('Category')['Sales'].sum().plot(kind='bar')
plt.show()
```

```python

# Q16. Line chart(monthly sales).
# Line chart
df.groupby('Month')['Sales'].sum().plot(kind='line',color='g') 
plt.show()
```

```python


# Q17. Pie char(region share).
# Pie chart
df.groupby('Region')['Sales'].sum().plot(kind='pie')
plt.show()
```
